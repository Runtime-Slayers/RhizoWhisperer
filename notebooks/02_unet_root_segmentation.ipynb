# RHIZO-NET: Notebook 02 - Root Segmentation Model Training
Trains and evaluates all 4 segmentation architectures:
1. **RhizoUNet** (Modified U-Net)
2. **RhizoAttentionNet** (Tubular Attention)
3. **DualStreamRootNet** (Spatial + Tubularity)
4. **RhizoHybridTransformer** (CNN-Transformer Hybrid)


```python
import torch
from src.unet.model import RhizoUNet
from src.unet.rhizo_attention_net import RhizoAttentionNet
from src.unet.dual_stream_root_net import DualStreamRootNet
from src.unet.rhizo_hybrid_transformer import RhizoHybridTransformer

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')

# Instantiate all models
models = {
    'RhizoUNet': RhizoUNet().to(device),
    'RhizoAttentionNet': RhizoAttentionNet().to(device),
    'DualStreamRootNet': DualStreamRootNet().to(device),
    'RhizoHybridTransformer': RhizoHybridTransformer(embed_dim=48, depth=2).to(device),
}

for name, model in models.items():
    params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f'{name}: {params:,} parameters')
```
